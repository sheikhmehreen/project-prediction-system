```python
import pandas as pd

cols = ["id", "project", "cat2", "forg", "center", "year", "mode",
        "rely", "data", "cplx", "time", "stor", "virt", "turn",
        "acap", "aexp", "pcap", "vexp", "lexp", "modp", "tool",
        "sced", "equivphyskloc", "act_effort"]

df = pd.read_csv("nasa93.csv", skiprows=25, header=None, names=cols)
print(df.shape)
print(df.isna().sum().sum())
df.head()
```

```python
print(df.shape)
print(df.isna().sum().sum())
print(df[["equivphyskloc", "act_effort"]].describe())
print(df["mode"].value_counts())
```

```python
a = df["mode"].map({"organic": 2.4, "semidetached": 3.0, "embedded": 3.6})
b = df["mode"].map({"organic": 1.05, "semidetached": 1.12, "embedded": 1.20})

df["estimate"] = a * df["equivphyskloc"] ** b
df["overrun"] = (df["act_effort"] > df["estimate"]).astype(int)

print(df["overrun"].value_counts())
```

# Project Overrun Predictor: Stage 1

**Dataset:** NASA93, a public software cost-estimation dataset. It has 93 projects and 24 columns, with no missing values.

**Overrun definition:** The dataset has no planned budget, so I calculate one. Budget = basic COCOMO estimate from project size (KLOC) and project mode. A project is an overrun (1) if its actual effort is higher than this budget, otherwise 0.

**Result:** 49 overruns and 44 non-overruns.

**Limits:** Only 93 projects, the budget is calculated and not a real planned figure, and only 3 projects are "organic" mode.
